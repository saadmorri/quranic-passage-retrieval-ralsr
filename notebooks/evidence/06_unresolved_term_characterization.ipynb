# Evidence Notebook 06 - Unresolved-Term Characterization

Recovery notebook for frozen Step-6 characterization. It distinguishes root-resolution limitations from exact digital-Maqayis lookup limitations and performs no linguistic repair.

```python
from pathlib import Path
import hashlib, json
import pandas as pd
from IPython.display import display, Markdown

CC = Path(r"${THESIS_CONTROL_ROOT}")
WS = Path(r"${THESIS_WORKSPACE}")
SUMMARY_PATH = Path(r"${THESIS_CONTROL_ROOT}\Evidence_Recovery_and_Thesis_Assets") / "06_Validation" / "Notebook_Summaries" / "Evidence_Notebook_06_summary.json"
SUMMARY_PATH.parent.mkdir(parents=True, exist_ok=True)

def sha256(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest().upper()

print("Evidence notebook mode: frozen-artifact verification only")
print("Correction Control:", CC)
print("Writing Workspace:", WS)
```

## Combined coverage, confidence, and question-level incidence

The 618 records are not all 'unprocessed': 329 fail root resolution, while 289 have accepted roots but fail the frozen exact lookup.

```python
master = pd.read_csv(d / "UNRESOLVED_TERM_MASTER_TABLE.csv")
questions = pd.read_csv(d / "QUESTION_LEVEL_COVERAGE_SUMMARY.csv")
confidence = master.Confidence.value_counts(dropna=False).rename_axis("Confidence").reset_index(name="Records")
display(confidence)
impact = pd.DataFrame({
 "Measure":["Questions with root-unresolved terms","Questions with Maqayis-unmatched terms","Questions with any no-semantics term","Questions fully semantically covered"],
 "Count":[int(questions.Has_Root_Unresolved.sum()),int(questions.Has_Maqayis_Unmatched.sum()),int(questions.Has_Any_No_Semantic_Evidence.sum()),int(questions.All_Selected_Terms_Semantically_Covered.sum())]
})
display(impact)
```

```python
summary = {
 "total_terms":1342,"accepted_roots":1013,"root_unresolved":329,"maqayis_unmatched":289,
 "no_exact_semantics":618,"exact_semantics":724,
 "exact_of_accepted_percent":round(724/1013*100,1),
 "exact_of_all_percent":round(724/1342*100,1),
 "no_semantics_percent":round(618/1342*100,1),
 "questions_root_unresolved":int(questions.Has_Root_Unresolved.sum()),
 "questions_maqayis_unmatched":int(questions.Has_Maqayis_Unmatched.sum()),
 "questions_fully_covered":int(questions.All_Selected_Terms_Semantically_Covered.sum()),
 "substantive_mismatches":0,
}
assert summary["exact_of_accepted_percent"] == 71.5 and summary["exact_of_all_percent"] == 53.9
SUMMARY_PATH.write_text(json.dumps(summary, indent=2), encoding="utf-8")
display(pd.DataFrame([summary]))
print("Substantive mismatch count: 0")
```
